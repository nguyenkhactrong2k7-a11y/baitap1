```python
# #W8A1
class Cyclinder:
    def __init__(self, radius, height):
        self.radius = radius
        self.height = height

    def Surface(self):
        self.surface = self.radius ** 2 * 3.14 * 2 + self.radius * self.height * 3.14 * 2
        return f"Surface = {self.surface:.2f}"

    def Volume(self):
        self.volume = self.radius ** 2 * 3.14 * self.height
        return f"Volume = {self.volume:.2f}"

radius, height = map(float, input().split())

result = Cyclinder(radius, height)
print(result.Surface())
print(result.Volume())
```

```python
#W8A2
class Date:
    def __init__(self, day, month, year):
        self.day, self.month, self.year = day, month, year

    def is_leap_year(self):
        return (self.year % 400 == 0) or (self.year % 4 == 0 and self.year % 100 != 0)

    def get_max_days(self):
        if self.month == 2 and self.is_leap_year():
            return 29
        return [0, 31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31][self.month]

    def is_valid_date(self):
        if not (1 <= self.month <= 12 and self.year >= 1):
            return False
        return 1 <= self.day <= self.get_max_days()

    def next_day(self):
        if self.day < self.get_max_days():
            self.day += 1
        else:
            self.day = 1
            if self.month < 12:
                self.month += 1
            else:
                self.month, self.year = 1, self.year + 1
        return self

    def __str__(self):
        return f"{self.day:02d}/{self.month:02d}/{self.year:04d}"

try:
    day, month, year = map(int, input().split('/'))
    date = Date(day, month, year)
    if not date.is_valid_date():
        print("INVALID")
    else:
        print(date.next_day())
except:
    print("INVALID")
```

```python
#W8A3
import math

class Point:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def is_origin(self):
        return self.x == 0 and self.y == 0

    def is_x_axis(self):
        return self.y == 0

    def is_y_axis(self):
        return self.x == 0

    def distance_to_origin(self):
        return math.sqrt(self.x**2 + self.y**2)

    def __str__(self):
        msg = ""
        if self.is_origin():
            msg = "La goc toa do \n"
        elif self.is_x_axis():
            msg = "Nam tren truc x \n"
        elif self.is_y_axis():
            msg = "Nam tren truc y \n"
            
        return f"{msg}Distance to origin: {self.distance_to_origin():.2f}"

x, y = map(float, input().split())
print(Point(x, y))
```

```python
#W8A4
class Calculator:
    def __init__(self, num1, num2):
        self.num1, self.num2 = num1, num2

    def add(self): return self.num1 + self.num2
    def subtract(self): return self.num1 - self.num2
    def product(self): return self.num1 * self.num2
    def power(self): return self.num1 ** self.num2

    def divide(self):
        return self.num1 / self.num2 if self.num2 != 0 else "Invalid operation"

    def mod(self):
        return self.num1 % self.num2 if self.num2 != 0 else "Invalid operation"

    def set_numbers(self, num1, num2):
        self.num1, self.num2 = num1, num2
        
num1, num2 = map(float, input("Enter two numbers: ").split())
calc = Calculator(num1, num2)

operations = {
    "1": calc.add,
    "2": calc.subtract,
    "3": calc.product,
    "4": calc.divide,
    "5": calc.power,
    "6": calc.mod
}

while True:
    print("\nMENU\n1. Add\n2. Subtract\n3. Product\n4. Divide\n5. Power\n6. Mod\n7. Set Numbers")
    choice = input("Choose operation: ")

    if choice in operations:
        print(f"Result: {operations[choice]()}")
    elif choice == "7":
        n1, n2 = map(float, input("Enter new numbers: ").split())
        calc.set_numbers(n1, n2)
    else:
        print("Invalid operation")

    if input("Do you want to exit? (yes/no): ").lower() == "yes":
        break
```

```python
#W8A5
class ShoppingCart:
    def __init__(self):
        self.items = []

    def add_item(self, name, price):
        self.items.append({'name': name, 'price': price})

    def delete_item(self, name):
        self.items = [i for i in self.items if i['name'] != name]

    def check_empty(self):
        return not self.items

    def total_price(self):
        return sum(item['price'] for item in self.items)

    def show_items(self):
        if self.items:
            print("\n".join(item['name'] for item in self.items))
        else:
            print("NONE")

    def clear(self):
        self.items.clear()

cart = ShoppingCart()
menu = "\nMENU\n2. Add\n3. Delete\n4. Empty?\n5. Total\n6. Show\n7. Clear"

while True:
    print(menu)
    choice = input("Choice: ")

    if choice == "2":
        cart.add_item(input("Name: "), float(input("Price: ")))
    elif choice == "3":
        cart.delete_item(input("Name: "))
    elif choice == "4":
        print("Is empty:", cart.check_empty())
    elif choice == "5":
        print("Total price:", cart.total_price())
    elif choice == "6":
        cart.show_items()
    elif choice == "7":
        cart.clear()

    if input("Exit? (yes/no): ").lower() == "yes":
        break
```

```python
#W8A6
class Rectangle:
    def __init__(self, width, height):
        self.width, self.height = width, height

    def scale(self, k):
        self.width *= k
        self.height *= k

    def area(self):
        return f"Area: {self.width * self.height:.2f}"

    def perimeter(self):
        return f"Perimeter: {2 * (self.width + self.height):.2f}"

w, h, k = map(float, input().split())
rect = Rectangle(w, h)
rect.scale(k)

print(rect.area())
print(rect.perimeter())
```

```python
#W8A7
import math

class Fraction:
    def __init__(self, num, den):
        self.num, self.den = num, den
        self._simplify()

    def _simplify(self):
        g = math.gcd(self.num, self.den)
        self.num //= g
        self.den //= g
        if self.den < 0:
            self.num, self.den = -self.num, -self.den

    def __add__(self, other):
        return Fraction(self.num * other.den + other.num * self.den, self.den * other.den)

    def __sub__(self, other):
        return Fraction(self.num * other.den - other.num * self.den, self.den * other.den)

    def __mul__(self, other):
        return Fraction(self.num * other.num, self.den * other.den)

    def __truediv__(self, other):
        return Fraction(self.num * other.den, self.den * other.num)

    def __str__(self):
        return f"{self.num}/{self.den}"

a, b, op, c, d = input().split()
f1, f2 = Fraction(int(a), int(b)), Fraction(int(c), int(d))

ops = {"+": f1 + f2, "-": f1 - f2, "*": f1 * f2, "/": f1 / f2}
print(ops[op])
```

```python
#W8A8
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount > self.balance:
            return 0
        else:
            self.balance -= amount

    def __str__(self):
        return f"{self.balance:.2f}"

owner, initial_balance = input().split()
initial_balance = float(initial_balance)
bank_account = BankAccount(owner, initial_balance)
n = int(input())
while n > 0:
    choice = input().split()
    if choice[0] == "DEPOSIT":
        bank_account.deposit(float(choice[1]))
    elif choice[0] == "WITHDRAW":
        bank_account.withdraw(float(choice[1]))
    n -= 1
print(str(bank_account))
```

```python
#W8A9
class Student:
    def __init__(self, name):
        self.name = name
        self.score = {}

    def add(self, subject, score):
        self.score[subject] = score

    def average(self):
        total = 0
        for score in self.score.values():
            total += score
        Average = total / len(self.score.keys())
        return Average

    def rank(self):
        if self.average() >=8:
            return "Excellent"
        elif self.average() < 8 and self.average() >=6.5:
            return "Good"
        elif 5 <= self.average() < 6.5:
            return "Average"
        else:
            return "Poor"

name = input()
student = Student(name)
n = int(input())
while n > 0:
    subject, score = input().split()
    student.add(subject, float(score))
    n-=1
average = student.average()

print(f"{name} {average:.2f} {student.rank()}")
```

```python
#W8A10
class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year

class Library:
    def __init__(self):
        self.books = []

    def add(self, book):
        self.books.append(book)

    def count_author(self, author):
        count = 0
        for book in self.books:
            if book.author == author:
                count += 1
        return count

    def count_year(self, year):
        count = 0
        for book in self.books:
            if book.year == year:
                count += 1
        return count

n = int(input())
library = Library()
while n > 0:
    choice = input()
    parts = choice.split()
    if parts[0] == "ADD":
        data = parts[1]
        info = data.split(';')
        title = info[0]
        author = info[1]
        year = int(info[2])
        book = Book(title, author, year)
        library.add(book)

    elif parts[0] == "COUNT":
        print(library.count_author(parts[1]))

    elif parts[0] == "COUNTYEAR":
        print(library.count_year(int(parts[1])))

    n-=1
```

```python
#W8A11
class Coordinate:
    def __init__(self, x, y, z):
        self.x = x
        self.y = y
        self.z = z

    def vector(self, other):
        x = other.x - self.x
        y = other.y - self.y
        z = other.z - self.z
        return Coordinate(x, y, z)

    def scalar_product(self, other):
        return (self.x * other.x + self.y * other.y + self.z * other.z)

    def cross_product(self, other):
        x = self.y * other.z - self.z * other.y
        y = self.z * other.x - self.x * other.z
        z = self.x * other.y - self.y * other.x
        return Coordinate(x, y, z)

    def length_vector(self):
        return math.sqrt(self.x**2 + self.y**2 + self.z**2)

xA, yA, zA = map(float, input().split())
xB, yB, zB = map(float, input().split())
xC, yC, zC = map(float, input().split())
xD, yD, zD = map(float, input().split())

A = Coordinate(xA, yA, zA)
B = Coordinate(xB, yB, zB)
C = Coordinate(xC, yC, zC)
D = Coordinate(xD, yD, zD)

AB = A.vector(B)
AC = A.vector(C)
cross_product_ABC = AB.cross_product(AC)
BC = B.vector(C)
BD = B.vector(D)
cross_product_BCD = BC.cross_product(BD)

num = cross_product_ABC.scalar_product(cross_product_BCD)
den = cross_product_ABC.length_vector() * cross_product_BCD.length_vector()
result = math.acos(abs(num / den))
result = math.degrees(result)
print(f"{result:.2f}")
```

```python
#W8A12
class ComplexNumber:
    def __init__(self, real, imaginary):
        self.real = real
        self.imaginary = imaginary

    def mul(self, other):
        real = self.real * other.real - self.imaginary * other.imaginary
        imaginary = self.real * other.imaginary + self.imaginary * other.real
        return ComplexNumber(real, imaginary)

    def __str__(self):
        return f"{self.real} {self.imaginary}"

realA, ImaginaryA = map(int, input().split())
realB, ImaginaryB = map(int, input().split())

A = ComplexNumber(realA, ImaginaryA)
B = ComplexNumber(realB, ImaginaryB)

result = A.mul(B)
print(str(result))
```

```python
#W8A13
class Triangle:
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c

    def is_valid(self):
        if self.a + self.b <= self.c or self.b + self.c <= self.a or self.c + self.a <= self.b:
            return False
        return True

    def area(self):
        p = (self.a + self.b + self.c) / 2
        result = math.sqrt(p*(p-self.a)*(p-self.b)*(p-self.c))
        return result

    def __str__(self):
        return f"{self.area():.2f}"

a, b, c = map(int, input().split())
triangle = Triangle(a, b, c)
if triangle.is_valid() == False:
    print("invalid")
else:
    if triangle.area() == int(triangle.area()):
        print(int(triangle.area()))
    else:
        print(str(triangle))

```
