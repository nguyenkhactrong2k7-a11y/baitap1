```python
#bai6
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
a, b = map(int, input().split())
while a>b:
    a, b = map(int, input().split())
x=0
for i in range(a, b+1):
    if is_prime(i):
        x+=i
print(x)
```

```python
#bai8
def reverse(n):
    i = 0
    s = 0
    if n < 0:
        n = abs(n)
        while n > 0:
            i = n % 10
            s = s * 10
            s = s + i
            n //= 10
        return -s
    while n > 0:
        i = n % 10
        s = s * 10
        s = s + i
        n //= 10
    return s
n = int(input())
print(reverse(n))
```

```python
#bai9
import math
try:
    x = int(input("Nhập giới hạn x: "))
except ValueError:
    print("Vui lòng nhập số nguyên!")
    exit()

def check_duplicate_bitwise(n):
    n = abs(n)
    if n > 9876543210:
        return False
    mask = 0
    while n > 0:
        digit = n % 10
        if (mask >> digit) & 1:
            return False
        mask |= (1 << digit)
        n //= 10
    return True 

print(f"Các số chính phương < {x} có các chữ số khác nhau là:")
h = 1
while True:
    sq = h * h  
    if sq >= x:
        break
    if check_duplicate_bitwise(sq):
        print(sq)
    h += 1
```

```python
#bai13
def nghiem(n):
    result = 0
    for i in range(1, n):
        if n % i == 0:
            result += i
    return result

a = int(input())
b = int(input())

if nghiem(a) == b and nghiem(b) == a:
    print("true")
else:
    print("false")
```
