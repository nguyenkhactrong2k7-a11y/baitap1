```python
#bai3
import math
def check_prime(n: int) -> bool:
    for i in range(2, math.isqrt(n)+1):
        if n% i ==0:
            return False
    return (n>=2)
```

```python
#bai4
n = int(input())
def kiemtrasohoanhao(n):
    s = 0
    for i in range(1, n):
        if n % i == 0:
            s = s + i
    if s == n:
        return True
    return False
if kiemtrasohoanhao(n):
    print(n,"la so hoan hao")
else:
    print(n,"khong la so hoan hao")
```

```python
#bai5
x,y = map(int, input().split())
def binary(n):
    arr = []
    while n > 0:
        arr.append(n % 2)
        n = n // 2
    arr2 = arr[::-1]
    return arr2
sum = 0
a = binary(x)
b = binary(y)
while len(a) < len(b):
    a.insert(0,0)
while len(b) < len(a):
    b.insert(0,0)
for i in range(len(a)):
    if a[i] != b[i]:
        sum +=1
print(sum)
```

```python
#bai6
def factorial(n):
    temp = 1
    if n == 0: return 0
    for i in range(1, n+1):
        temp *= i
    return temp
n = int(input())
print(factorial(n))
```

```python
#bai7
def casio(num1, num2, operator):
    if operator == "-":
        return f"{(num1 - num2):.2f}"
    elif operator == "+":
        return f"{(num1 + num2):.2f}"
    elif operator == "*":
        return f"{(num1 * num2):.2f}"
    elif operator == "/":
        return f"{(num1 / num2):.2f}"
num1, num2 = map(float, input().split())
operator = input()
print(casio(num1, num2, operator))
```

```python
#bai9
def sum_digit(num):
    sum = 0
    while num > 0:
        sum += num % 10
        num //= 10
    return sum
n = int(input())
print(sum_digit(n))
```
