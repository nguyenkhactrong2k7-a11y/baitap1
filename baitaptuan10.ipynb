```python
#W9A1
lst1 = list(map(int, input().split()))
lst2 = list(map(int, input().split()))
merge_lst = []
i, j = 0, 0
while i < len(lst1) and j < len(lst2):
    if lst1[i] < lst2[j]:
        merge_lst.append(lst1[i])
        i += 1
    else:
        merge_lst.append(lst2[j])
        j += 1
merge_lst.extend(lst1[i:])
merge_lst.extend(lst2[j:])
print(*merge_lst)
```

```python
#W9A2
matrix = []
while True:
    try:
        line = input().strip()
        if not line: break
        matrix.append(list(map(int, line.split())))
    except EOFError:
        break

n = int(matrix.pop()[0])
flat_list = []
for row in matrix:
    for val in row:
        flat_list.append(val)

if 1 <= n <= len(flat_list):
    print(flat_list[n-1])
```

```python
#W9A3
number = list(input().strip())
swap = []
idx_min = 0
res = ""

for i in range(len(number)):
    min_val = "10"
    for j in range(len(number) - 1, i, -1):
        if number[j] < number[i] and number[j] < min_val:
            min_val = number[j]

    if min_val != "10":
        for k in range(len(number) - 1, i, -1):
            if number[k] == min_val:
                idx_min = k
                break
        swap.extend([i, idx_min])
        break

if not swap:
    if len(number) >= 2:
        number[-1], number[-2] = number[-2], number[-1]
else:
    number[swap[0]], number[swap[1]] = number[swap[1]], number[swap[0]]

for i in range(len(number)):
    res += str(number[i])
print(int(res))
```

```python
#W9A4
def check():
    a = input().strip()
    b = input().strip()
    n,m = len(a),len(b)
    dp = []
    for i in range(n+1):
        row = []
        for j in range(m+1):
            row.append(False)
        dp.append(row)

    dp[0][0] = True
    for i in range(n):
        for j in range(m+1):
            if dp[i][j] == True:
                if j<m and a[i].upper()==b[j]:
                    dp[i+1][j+1] = True
                if a[i].islower():
                    dp[i+1][j] = True

    if dp[n][m] == True:
        print("YES")
    else:
        print("NO")

n = int(input())
for i in range(n):
    check()
```

```python
#W9A5 
def is_subarray(list1, list2):
    n, m = len(list1), len(list2)
    if n > m:
        return False
    for i in range(m - n + 1):
        if list2[i:i+n] == list1:
            return True
    return False
list1 = list(map(int, input().split()))
list2 = list(map(int, input().split()))
if is_subarray(list1, list2):
    print("YES")
else:
    print("NO")
```

```python
#W9A6 
def next_permutation(n, arr):
    i = n - 2
    while i >= 0 and arr[i] >= arr[i+1]:
        i -= 1
        
    if i == -1:
        return sorted(arr)
        
    j = n - 1
    while arr[j] <= arr[i]:
        j -= 1
        
    arr[i], arr[j] = arr[j], arr[i]
    arr[i+1:] = arr[i+1:][::-1]
    return arr

try:
    n = int(input())
    arr = list(map(int, input().split()))
    print(*(next_permutation(n, arr)))
except EOFError:
    pass
```

```python
#W9A7 
def canPlaceFlowers(flowerbed, k):
    n = len(flowerbed)
    for i in range(n):
        if flowerbed[i] == 0:
            left = (i == 0) or (flowerbed[i-1] == 0)
            right = (i == n-1) or (flowerbed[i+1] == 0)
            if left and right:
                flowerbed[i] = 1
                k -= 1
                if k == 0:
                    return True
    return k <= 0
flowerbed = list(map(int, input().split()))
k = int(input())
print(canPlaceFlowers(flowerbed, k))
```

```python
#W9A8
def is_path_crossing(moves: str) -> bool:
    x, y = 0, 0
    visited = {(0, 0)}
    for move in moves:
        if move == 'R':
            x += 1
        elif move == 'L':
            x -= 1
        elif move == 'U':
            y += 1
        elif move == 'D':
            y -= 1
        if (x, y) in visited:
            return True
        visited.add((x, y))
    return False
moves = input().strip()
print(is_path_crossing(moves))
```

```python
#W9A9 
def Median(nums):
    a, b, c, d, e = nums
    if a > b: a, b = b, a
    if c > d: c, d = d, c
    if a > c: a, c = c, a
    if b > d: b, d = d, b
    if b > c: b, c = c, b
    if c > e: c, e = e, c
    return c
nums = list(map(int, input().split()))
print(Median(nums))
```

```python
#W9A10
import bisect

def findRadius(houses, heaters):
    houses.sort()
    heaters.sort()
    res = 0
    for h in houses:
        
        idx = bisect.bisect_left(heaters, h)
        
        left = h - heaters[idx-1] if idx > 0 else float('inf')
        
        right = heaters[idx] - h if idx < len(heaters) else float('inf')

        nearest = min(left, right)
        res = max(res, nearest)
    return res
houses = list(map(int, input().split()))
heaters = list(map(int, input().split()))
n = int(input())
print(findRadius(houses, heaters))
```
