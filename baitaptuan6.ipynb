```python
#bai1
chuoi = list(map(int, input().split()))
chuoi1 = []
for so in chuoi:
    if so not in chuoi1:
        chuoi1.append(so)
print(chuoi1)
```

```python
#bai2
chuoi = list(map(int, input().split()))
tichluy = []
def tichluy1(x):
    tong = 0
    for i in range(0, x + 1):
        tong = tong + chuoi[i]
    return tong
for i in range(0, len(chuoi)):
    tichluy.append(tichluy1(i))
print(tichluy)
```
