```python
#W7A1
def binary_search(arr, left, right, target):
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        elif arr[mid] > target:
            right = mid - 1
    return '-1'

lst = list(map(int, input().split()))
k = int(input())
print(binary_search(lst, 0, len(lst)-1, k))
```

```python
#W7A3
def bubble_sort(arr):
    for i in range(len(arr)):
        for j in range(len(arr)-i-1):
            if arr[j]> arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
    return arr

def selection_sort(arr):
    for i in range(len(arr)):
        min_index = i
        for j in range(i+1, len(arr)):
            if arr[min_index] > arr[j]:
                min_index = j
        arr[i], arr[min_index] = arr[min_index], arr[i]
    return arr

lst = list(map(int, input().split()))
print(bubble_sort(lst))
print(selection_sort(lst))
```

```python
#W7A5
def count_pair(lst, k):
    fre = {}
    count = 0

    for x in lst:
        y = k - x
        if y in fre:
            count += fre[y]
        if x not in fre:
            fre[x] = 1
        else:
            fre[x] += 1

    return count

lst = list(map(int, input().split()))
k = int(input())
print(count_pair(lst, k))
```

```python
#W7A8
def shortest_interval(arr, k):
    temp = []
    interval = []
    for i in range(len(arr)):
        if arr[i][0] <= k and k <= arr[i][1]:
            temp.append(arr[i])

    if temp==[]:
        return -1

    for i in range(len(temp)):
            interval.append(temp[i][1]-temp[i][0]+1)
    return min(interval)

lst = eval(input())
queries = eval(input())
storage = []
for i in queries:
    storage.append(shortest_interval(lst, i))
print(storage)
```

```python
#W7A9
def siu(arr):
    ans=[]
    for i in range(len(arr)):
        dem=0
        for j in range(i+1, len(arr)):
            if arr[j] < arr[i]:
                dem+=1
        ans.append(dem)
    return ans
raw_input = input().strip()
clean_input = raw_input.replace('[', '').replace(']', '')
A = [int(x) for x in clean_input.split(',') if x.strip()]

print(siu(A))
```

```python
#W7A10
import bisect

def tong_mang_con_lon_nhat_mod_m(nums, m):
    prefix = 0
    ds_prefix = []
    ket_qua = 0

    for x in nums:
        prefix = (prefix + x) % m

        ket_qua = max(ket_qua, prefix)

        i = bisect.bisect_right(ds_prefix, prefix)
        if i < len(ds_prefix):
            ket_qua = max(ket_qua, (prefix - ds_prefix[i] + m) % m)

        bisect.insort(ds_prefix, prefix)

    return ket_qua

nums=[7, 11, 14, 2, 8, 4]
m=12
print(tong_mang_con_lon_nhat_mod_m(nums, m))
```
